```python
# 30.1 위치인수
def print_numbers(a, b, c):
    print(a)
    print(b)
    print(c)
```

```python
print_numbers(10,20,30)
```

```python
# 30.1.2 언패킹 사용하기

x = [10, 20, 30]
print_numbers(*[10, 20, 30])
```

```python
# 30.2 키워드 인수

def personal_info(name, age, address):
    print("이름:", name)
    print("나이:", age)
    print("주소:", address)
```

```python
personal_info(age=30, address="서울시", name="홍길동")
```

```python
x = {
    "name": "홍길동", 
    "age": 30, 
    "address": "서울시 용산구"
}

personal_info(**x)
personal_info(name="홍길동", age=30, address="서울시 용산구")
```

```python
# 가변 키워드 인수 함수 만들기
# def 함수이름(**매개변수): 
#     코드

def personal_info(**kwargs):
    # print(kwargs)
    for kw, arg in kwargs.items():
        print(kw, arg)
```

```python
personal_info(name="홍길동", age=30, address="서울시 용산구")
```

```python
# 매개변수에 초기값 지정하기

def personal_info(name, age, address="비공개"):
    print("이름:", name)
    print("나이:", age)
    print("주소:", address)
```

```python
personal_info("홍길동",30)
```

```python
personal_info("홍길동",30, "서울시 용산구")
```
