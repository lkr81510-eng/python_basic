```python
# 리스트에 요소 하나 추가하기 

a = [10, 20, 30]
a.append(500)

print(a)
```

```python
a = []
a.append(10)
print(a)
```

```python
# 리스트 안에 리스트 추가하기 
a = [10, 20, 30]

a.append([500, 600])
print(a)
```

```python
len(a)
```

```python
# 리스트 확장하기 
a = [10, 20, 30]
a.extend([500, 600])
print(a)
```

```python
len(a)
```

```python
# 리스트의 특정 인덱스에 요소 추가하기 
a = [10, 20, 30]
a.insert(2, 500)
print(a)
```

```python
a = [10, 20, 30]
a.insert(0, 500)
print(a)
```

```python
# 리스트에서 요소 삭제하기

a = [10, 20, 30]
a.pop()
print(a)
```

```python
a = [10, 20, 30]
a.pop()
```

```python
# 리스트에서 특정 값을 찾아서 삭제하기 
a = [10, 20, 30]
a.remove(20)
print(a)
```

```python
# 리스트에서 특정 값의 인덱스 구하기
a = [10, 20, 30, 15, 20, 40]
a.index(20)
```

```python
# 특정 값의 개수 구하기 
a = [10, 20, 30, 15, 20, 40]
a.count(20)
```

```python
# 리스트의 순서를 뒤집기
# reversed()
a = [10, 20, 30, 15, 20, 40]
a.reverse()
a
```

```python
# 리스트의 요소 정렬하기
a = [10, 20, 30, 15, 20, 40]
a.sort()
print(a)
```

```python
a = [10, 20, 30, 15, 20, 40]
a.sort(reverse=True)
```
